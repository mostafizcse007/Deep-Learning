```python
import numpy as np 
import pandas as pd 

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report,r2_score,mean_absolute_error

from tensorflow.keras.layers import Dense,Layer,Input
from keras import Sequential
```

```python
df = pd.read_csv('Admission_Predict.csv')
```

```python
df.head()
```

```python
df.drop(columns=['Serial No.'],inplace=True)
```

```python
X = df.iloc[:,0:7]
y = df.iloc[:,-1]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2,random_state=1)
```

```python
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model = Sequential()
model.add(Input(shape=(X_train_scaled.shape[1],)))
```

```python
model.add(Dense(7,activation='relu'))
model.add(Dense(7,activation='relu'))
model.add(Dense(1,activation='linear'))
```

```python
model.summary()
```

```python
model.compile(loss='mean_squared_error',optimizer='Adam',metrics=['r2_score'])
```

```python
history = model.fit(x=X_train_scaled,y=y_train,epochs=1000,validation_split=.2)
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
r2_score(y_test,y_pred)
```
