```python
a = torch.tensor(3.0,requires_grad=True)
```

```python
b = a**2
```

```python
b.backward()
```

```python
a.grad
```

```python
x = torch.tensor(3.0,requires_grad=True)
y = x**2
z = torch.sin(y)
```

```python
z.backward()
```

```python
x.grad
```

##### Multi dimentional derivatives

```python
x = torch.tensor([1,2,3],requires_grad=True,dtype=torch.float32)
```

```python
y = (5*x**2 + 3*torch.sin(x)).mean()
```

```python
y.backward()
```

```python
x.grad
```

```python
x.grad.zero_() # replacing previous stored values with zeros
```

##### Stopping backward pass

```python
x = torch.tensor(2.0,requires_grad=True)
```

```python
y = x**2
```

```python
y.backward()
```

```python
x.grad
```
