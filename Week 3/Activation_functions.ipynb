```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-10,10,10000)
```

```python
def sigmoid(x):
  return 1/(1+np.exp(-x))
```

```python
y = sigmoid(x)
```

# Sigmoid

```python
plt.plot(x,y)
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.show()
```

```python
def tanh(x):
  num = np.exp(x) - np.exp(-x)
  den = np.exp(x) + np.exp(-x)
  return num/den
```

```python
y = tanh(x)
```

# tanh

```python
plt.plot(x,y)
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.show()
```

```python
def relu(x):
  return np.maximum(0,x)
```

# Relu

```python
y = relu(x)
```

```python
plt.plot(x,y)
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.show()
```

```python
def leaky_relu(x):
  return np.maximum(x,0.1*x)
```

```python
y = leaky_relu(x)
```

# Leaky Relu

```python
plt.plot(x,y)
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.show()
```

```python
def elu(x,alpha=0.01):
  return np.where(x>0,x,alpha*(np.exp(x)-1))
```

```python
y = elu(x,.3)
```

# Elu

```python
plt.plot(x,y)
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.show()
```

```python
def selu(x,alpha=1.6732,lambdaa=1.0507):
  return lambdaa*np.where(x>0,x,alpha*(np.exp(x)-1))
```

```python
y = selu(x)
```

# Selu

```python
plt.plot(x,y)
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.show()
```
