```python
import re
```

##### Character set 

```python
txt = "The rain spain"
x = re.findall(r'[a-m]',txt)
x
```

##### Escape character

```python
txt = "Hi this is Antu, My age is 24"
x = re.findall(r'\w+',txt)
x
```

##### Digit character class

```python
txt = "that will be 1 105 008 dollars"
x = re.findall(r'\d{2,3}+',txt)
x
```

##### At the beginning

```python
txt = "123 123 this is my number 01309462705"
x = re.findall(r'^\d+',txt)
x
```

##### At the end

```python
txt = "123 123 this is my number 01309462705"
x = re.findall(r'\d+$',txt)
x
```

##### Wildcard 

```python
txt = "hello world"
x = re.findall(r'he..o',txt)
x
```

##### Asterics

```python
txt = "juniormostafiz@gmail.com"
x = re.findall(r'.*@.*\.*',txt)
x
```

##### Ends with

```python
txt = "colorful beautiful wonderful"
x = re.findall(r'.*ful+$',txt)
x
```
