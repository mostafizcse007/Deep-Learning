###### Removing html tags

```python
import re
def remove_html(text):
    pattern = re.compile(r'<.*?>')
    return pattern.sub(r'',text)
```

```python
df['review'] = df['review'].apply(remove_html)
```

###### Remove url

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
```

```python
df['review'] = df['review'].apply(remove_url)
```

```python
def text_conversion(text):
    text = text.lower()
    new_text = []
    for w in text.split():
        if w in abbreviations:
            new_text.append(abbreviations[w])
        else:
            new_text.append(w)
    return " ".join(new_text)
```

###### Stopwords removal

```python
import nltk
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
def remove_stopwords(text):
    text = text.lower()
    words = text.split()
    cleaned = []
    for word in words:
        if word and word not in stop_words:
            cleaned.append(word)
    return " ".join(cleaned)
```

```python
remove_stopwords("Hi this is Antu")
```

###### Remove emojies

```python
def remove_emoji(text):
    emoji_pattern = re.compile(
    "["
    "\U0001F600-\U0001F64F"  # Emoticons
    "\U0001F300-\U0001F5FF"  # Symbols & pictographs
    "\U0001F680-\U0001F6FF"  # Transport & map symbols
    "\U0001F1E0-\U0001F1FF"  # Flags
    "\U00002702-\U000027B0"
    "\U000024C2-\U0001F251"
    "]+",
    flags=re.UNICODE
)
    return emoji_pattern.sub("",text)
```

```python
remove_emoji("I love Bangladesh 🔥❤️💖")
```

###### Stemming

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
stem_words("having a sharp memory is kind of blessing")
```

###### Lemmitization

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
```

```python
lemma = WordNetLemmatizer()
sentence = "He was running and eating at the same time"
```

```python
print(f"Original{'-'*20}Lemmatize")
for word in sentence.split():
    ans = lemma.lemmatize(word,pos='v')
    print(f"{word}{'-'*20}{ans}")
    
```
